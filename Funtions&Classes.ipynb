```python
class person:

  def __init__(self,name,gender,age):
    self.name=name
    self.gender=gender
    self.age=age

  def talk(self):
    print("hi I am ",self.name)

  def vote(self):
    if self.age<18:
      print("I am not eligible to vote")
    else:
      print("I am eligible to vote")
obj=person("punom","female",22)
obj.talk()
obj.vote()
obj=person("Nafis","male",24)
obj.talk()
obj.vote()

```
